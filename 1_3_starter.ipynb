### 🔍 **Key Insights:**

- **Confusion matrices** reveal per-class performance that overall accuracy alone can mask
- **Precision vs Recall tradeoff**: High precision means fewer false positives, high recall means fewer false negatives
- **F1-score** provides a balanced measure when classes may have different error patterns
- Setting `shuffle=False` on the test generator is critical for correct evaluation alignment
